# Load the data

Place the files in the ../../05_src/data/fires/ directory and load the appropriate file. 

```python
# Load the libraries as required.

import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, train_test_split
```

```python
# Load data
columns = [
    'coord_x', 'coord_y', 'month', 'day', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain', 'area' 
]
# Had to update the code below to load
fires_dt = pd.read_csv(r"..\..\05_src\data\fires\forestfires.csv", header=0, names=columns)
fires_dt.info()
```

# Get X and Y

Create the features data frame and target data.

```python
# Define the features by removing the 'area' column
X = fires_dt.drop(columns=["area"])

# Display the first few rows
print(X.head())
```

```python
Y = fires_dt["area"]

# Define features
num_features = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']
cat_features = ['month', 'day']
```

# Preprocessing

Create two [Column Transformers](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html), called preproc1 and preproc2, with the following guidelines:

- Numerical variables

    * (Preproc 1 and 2) Scaling: use a scaling method of your choice (Standard, Robust, Min-Max). 
    * Preproc 2 only: 
        
        + Choose a transformation for any of your input variables (or several of them). Evaluate if this transformation is convenient.
        + The choice of scaler is up to you.

- Categorical variables: 
    
    * (Preproc 1 and 2) Apply [one-hot encoding](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) where appropriate.


+ The only difference between preproc1 and preproc2 is the non-linear transformation of the numerical variables.
    


### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python

preproc1 = ColumnTransformer(transformers=[
    ('num', StandardScaler(), num_features),  
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
])
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
numeric_transformer_2 = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('power', PowerTransformer())  # Applying Power Transformation
])

preproc2 = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_2, num_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ]
)
```

```python
#Print Transformers
print(preproc1)
print(preproc2)
```

## Model Pipeline


Create a [model pipeline](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html): 

+ Add a step labelled `preprocessing` and assign the Column Transformer from the previous section.
+ Add a step labelled `regressor` and assign a regression model to it. 

## Regressor

+ Use a regression model to perform a prediction. 

    - Choose a baseline regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Choose a more advance regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Both model choices are up to you, feel free to experiment.

```python
# Pipeline A = preproc1 + baseline
pipeline_A = Pipeline(steps=[
    ('preprocessing', preproc1),  # Preprocessing step
    ('regressor', Ridge())        # Baseline regressor (Ridge)
])

pipeline_A
```

```python
# Pipeline B = preproc2 + baseline
pipeline_B = Pipeline(steps=[
    ('preprocessing', preproc2),  # Preprocessing step
    ('regressor', Ridge())        # Baseline regressor (Ridge)
])

pipeline_B
```

```python
# Pipeline C = preproc1 + advanced model
pipeline_C = Pipeline(steps=[
    ('preprocessing', preproc1),  # Preprocessing step
    ('regressor', RandomForestRegressor(random_state=42))  # Advanced regressor (RandomForest with set random seed for reproducibility)
])

pipeline_C
```

```python
# Pipeline D = preproc2 + advanced model
pipeline_D = Pipeline(steps=[
    ('preprocessing', preproc2),  # Preprocessing step
    ('regressor', RandomForestRegressor(random_state=42))  # Advanced regressor (RandomForest)
])

pipeline_D
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
# Hyperparameter grid for Ridge (baseline model)
ridge_params = {
    'regressor__alpha': [0.01, 0.1, 1, 10]
}
```

```python
# Hyperparameter grid for RandomForestRegressor (advanced model)

rf_params = {
    'regressor__n_estimators': [50, 100, 150, 200],  
    'regressor__max_depth': [None, 10, 20, 30]       
}
```

```python
# GridSearch setup for each pipeline and fit each GridSearchCV object

grid_a = GridSearchCV(pipeline_A, param_grid=ridge_params, cv=5, scoring='neg_mean_squared_error')
grid_a.fit(X, Y)

grid_b = GridSearchCV(pipeline_B, param_grid=ridge_params, cv=5, scoring='neg_mean_squared_error')
grid_b.fit(X, Y)

grid_c = GridSearchCV(pipeline_C, param_grid=rf_params, cv=5, scoring='neg_mean_squared_error')
grid_c.fit(X, Y)

grid_d = GridSearchCV(pipeline_D, param_grid=rf_params, cv=5, scoring='neg_mean_squared_error')
grid_d.fit(X, Y)
```

```python
# Collect the best scores and parameters
print("Pipeline A - Best Score:", -grid_a.best_score_, "Best Params:", grid_a.best_params_)
print("Pipeline B - Best Score:", -grid_b.best_score_, "Best Params:", grid_b.best_params_)
print("Pipeline C - Best Score:", -grid_c.best_score_, "Best Params:", grid_c.best_params_)
print("Pipeline D - Best Score:", -grid_d.best_score_, "Best Params:", grid_d.best_params_)
```

```python
scores = {
    "Pipeline A": -grid_a.best_score_,
    "Pipeline B": -grid_b.best_score_,
    "Pipeline C": -grid_c.best_score_,
    "Pipeline D": -grid_d.best_score_,
}

# Identify the best performing pipeline
best_pipeline = min(scores, key=scores.get)
best_model_score = scores[best_pipeline]

# Display the results
print(f"The best performing model is {best_pipeline} with a score of {best_model_score:.4f}")
```

+ Save the best performing model to a pickle file.

```python
import pickle
# Get the best performing model from grid search
best_performing_model = grid_a.best_estimator_

# Save the best model to a file
with open('best_best_performing_model.pkl', 'wb') as file:
    pickle.dump(best_performing_model, file)
```

# Explain

+ Use SHAP values to explain the following only for the best-performing model:

    - Select an observation in your test set and explain which are the most important features that explain that observation's specific prediction.

    - In general, across the complete training set, which features are the most and least important.

+ If you were to remove features from the model, which ones would you remove? Why? How would you test that these features are actually enhancing model performance?

```python
import shap

# Fit the best performing model to the training data
best_performing_model.fit(X_train, Y_train)

# Transform the test data using the preprocessing step from the pipeline
data_transform = best_performing_model.named_steps['preprocessing'].transform(X_test)

# Initialize the SHAP explainer for a linear model
explainer = shap.explainers.Linear(
    best_performing_model.named_steps['regressor'],
    data_transform,
    feature_names=best_performing_model.named_steps['preprocessing'].get_feature_names_out()
)

# Calculate SHAP values for the transformed data
shap_values = explainer(data_transform)
```

```python
shap.summary_plot(shap_values, data_transform)
shap.plots.waterfall(shap_values[1])
```

Positive Contributors: The most important features contributing positively to the prediction are dmc (8.94) and coord_x (6.08).

Negative Contributors: The features dc (-2.60), rain (-2.40), and rh (-2.36) are the main negative contributors. 

Base on the result from SHAP analysis, I would remove features with SHAP values close to zero, as these have less impact to predictions.  For example, day_fri, wind, month_sept. To verify this, I would retrain the model excluding these features and evaluate its performance using cross-validation
